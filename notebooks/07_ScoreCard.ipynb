```python
import os
import sys

from   IPython.display      import HTML, Image, display

sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))

from   src.notebook_outputs import (
                                   SCORECARD_ANALYSES,
                                   figure_label,
                                   list_notebook_figures,
                                   notebook_outputs_dir,
                                  )
```

```python
outputs_dir = input('Path to saved notebook figures') or '../data/notebook_outputs'
outputs_dir = os.path.abspath(outputs_dir)
```

```python
DASHBOARD_SECTIONS = [
    {
        'id'         : 'congruence',
        'label'      : 'Congruence',
        'notebook'   : '01_Congruence.ipynb',
        'step_class' : 'congruence',
        'description': 'Statistical similarity between real and synthetic patch feature distributions.',
    },
    {
        'id'         : 'coverage',
        'label'      : 'Coverage',
        'notebook'   : '02_Coverage.ipynb',
        'step_class' : 'coverage',
        'description': 'Extent to which synthetic data covers the real feature space.',
    },
    {
        'id'         : 'completeness',
        'label'      : 'Completeness',
        'notebook'   : '03_Completeness.ipynb',
        'step_class' : 'completeness',
        'description': 'Missing values and required-field fill rates in metadata CSVs.',
    },
    {
        'id'         : 'consistency',
        'label'      : 'Consistency',
        'notebook'   : '04_Consistency.ipynb',
        'step_class' : 'consistency',
        'description': 'Stability of numeric metadata across demographic subgroups.',
    },
    {
        'id'         : 'constraint',
        'label'      : 'Constraint',
        'notebook'   : '05_Constraint.ipynb',
        'step_class' : 'constraint',
        'description': 'Rate at which synthetic values violate real-data percentile bounds.',
    },
]

SECTION_STYLES = {
    'congruence'  : ('#7c3aed', 'white'),
    'coverage'    : ('#2563eb', 'white'),
    'completeness': ('#16a34a', 'white'),
    'consistency' : ('#ea580c', 'white'),
    'constraint'  : ('#eab308', '#422006'),
}
```

```python
def dashboard_title():
    display(HTML(
        """
        <div style="margin:0 0 1.5rem;padding:1.5rem;border-radius:16px;
                    background:linear-gradient(135deg,#eef2ff 0%,#f8fafc 100%);
                    border:1px solid #e2e8f0;">
            <h2 style="margin:0 0 0.5rem;color:#1e293b;">ScoreCard Dashboard</h2>
            <p style="margin:0;color:#475569;">
                Saved figures from notebooks 01–05, excluding Comprehension and Histograms.
            </p>
        </div>
        """
    ))


def section_header(section):
    bg, fg = SECTION_STYLES.get(section['step_class'], ('#64748b', 'white'))
    display(HTML(
        f"""
        <div style="margin:2.5rem 0 1rem;padding:1.25rem 1.5rem;border:1px solid #e2e8f0;
                    border-radius:12px;background:#f8fafc;">
            <div style="display:flex;align-items:center;gap:1rem;flex-wrap:wrap;">
                <span style="font-size:0.8rem;font-weight:700;text-transform:uppercase;
                             letter-spacing:0.75px;background:{bg};color:{fg};
                             padding:6px 18px;border-radius:9999px;">{section['label']}</span>
                <span style="font-size:1.15rem;font-weight:600;color:#1e293b;">{section['label']} Results</span>
                <span style="font-size:0.85rem;color:#64748b;">from {section['notebook']}</span>
            </div>
            <p style="margin:0.75rem 0 0;color:#475569;font-size:0.95rem;line-height:1.5;">{section['description']}</p>
        </div>
        """
    ))


def missing_section_notice(section):
    display(HTML(
        f"""
        <div style="margin:0 0 1.5rem;padding:1rem 1.25rem;border:1px dashed #cbd5e1;
                    border-radius:10px;background:#fff;color:#64748b;">
            No saved figures found for <strong>{section['label']}</strong>.
            Run <code>{section['notebook']}</code> first.
        </div>
        """
    ))


def display_saved_figure(path, analysis_id):
    label = figure_label(path, analysis_id)
    display(HTML(
        f"""
        <div style="margin:0 0 1.75rem;padding:1rem 1rem 0.5rem;border:1px solid #e2e8f0;
                    border-radius:12px;background:white;box-shadow:0 2px 8px rgba(0,0,0,0.05);">
            <div style="margin:0 0 0.75rem;font-size:0.95rem;font-weight:600;color:#334155;">{label}</div>
        </div>
        """
    ))
    display(Image(filename=path, embed=True))


def render_section(section, base_dir):
    section_header(section)
    paths = list_notebook_figures(section['id'], base_dir=base_dir)

    if not paths:
        missing_section_notice(section)
        return 0

    for path in paths:
        display_saved_figure(path, section['id'])

    return len(paths)
```

```python
notebook_outputs_dir(outputs_dir)
dashboard_title()

section_counts = []
for section in DASHBOARD_SECTIONS:
    count = render_section(section, outputs_dir)
    section_counts.append({
        'Analysis' : section['label'],
        'Notebook' : section['notebook'],
        'Figures'  : count,
        'Status'   : 'Ready' if count else 'Missing — run notebook first',
    })

display(HTML('<h3 style="margin:2rem 0 0.75rem;color:#1e293b;">Dashboard Status</h3>'))

rows = ''.join(
    f"<tr><td style='padding:8px 12px;border-bottom:1px solid #e2e8f0;'>{row['Analysis']}</td>"
    f"<td style='padding:8px 12px;border-bottom:1px solid #e2e8f0;'><code>{row['Notebook']}</code></td>"
    f"<td style='padding:8px 12px;border-bottom:1px solid #e2e8f0;text-align:center;'>{row['Figures']}</td>"
    f"<td style='padding:8px 12px;border-bottom:1px solid #e2e8f0;'>{row['Status']}</td></tr>"
    for row in section_counts
)

display(HTML(
    f"""
    <table style="border-collapse:collapse;width:100%;max-width:900px;
                  background:white;border:1px solid #e2e8f0;border-radius:12px;overflow:hidden;">
        <thead>
            <tr style="background:#f8fafc;">
                <th style="padding:10px 12px;text-align:left;color:#475569;">Analysis</th>
                <th style="padding:10px 12px;text-align:left;color:#475569;">Notebook</th>
                <th style="padding:10px 12px;text-align:center;color:#475569;">Figures</th>
                <th style="padding:10px 12px;text-align:left;color:#475569;">Status</th>
            </tr>
        </thead>
        <tbody>{rows}</tbody>
    </table>
    <p style="margin:0.75rem 0 0;color:#64748b;font-size:0.9rem;">
        Figures are read from <code>{outputs_dir}</code>
    </p>
    """
))
```

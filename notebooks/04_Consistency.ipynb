```python
import os
import sys
```

```python
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))

from src.analyses import consistency_analysis
from src.notebook_outputs import save_notebook_figures
```

```python
fields_to_plot = {
    'Age at dx'            : True,
    'BMI at dx (kg)'       : True,
    'BMI at follow-up (kg)': True,
    'appmag'               : True,
    'exposure scale'       : True,
    'exposure time'        : True,
    'focus offset'         : True,
    'left'                 : True,
    'lineareaxoffset'      : True,
    'lineareayoffset'      : True,
    'linecameraskew'       : True,
    'mpp'                  : True,
    'stripewidth'          : True,
    'top'                  : True,
    'levels'               : True,
    'tileheight'           : True,
    'tilewidth'            : True,
    'compressionratio'     : True,
}

metric_cols = [field for field, enabled in fields_to_plot.items() if enabled]
group_by    = 'Race'
```

```python
metrics_to_plot = {
    'Variance_of_Group_Means': True,
    'Max_Min_Difference'     : True,
    'ANOVA_F_statistic'      : True,
}
```

```python
real_meta_csv  = '../data/real_patch_appearance_with_metadata.csv'
synth_meta_csv = '../data/kde_patch_appearance_with_metadata.csv'

cons_df, cons_figs = consistency_analysis(
    real_meta_csv,
    synth_meta_csv,
    group_by=group_by,
    metric_cols=metric_cols,
    label="report",
    metrics_to_plot=metrics_to_plot,
)

if cons_figs:
    for fig in cons_figs.values():
        fig.show()

consistency_prefix = f"consistency_{group_by.replace(' ', '_')}"
saved_paths = save_notebook_figures(
    'consistency',
    cons_figs,
    prefix=consistency_prefix,
)
print(f'Saved {len(saved_paths)} figure(s) for the ScoreCard dashboard.')

```

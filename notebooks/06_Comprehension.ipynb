**Comprehension** Checklist to evaluate the transparency and clarity of the data generation process as well as the quality of the accompanying documentation.

- File accompanying every synthetic dataset (JSON, YAML, or Markdown).


### Generation
- `generation_method` — Method used to generate the data (e.g., GAN, Diffusion, VAE, Rule-based, KDE, etc.)
- `model_version` / `checkpoint` — Specific model iteration
- `hyperparameters` — Key training and generation hyperparameters (as a structured object or string)
- `random_seed` — Random seed used for reproducibility

### Input
- `real_data_source` — Description or reference to the real dataset(s) used for training/generation
- `preprocessing_steps` — List of preprocessing, augmentation, or normalization steps applied

### Evaluation
- `initial_evaluation` — e.g. internal verification during model development or other down-stream task (non-7Cs) evaluation

### Usage
- `intended_use` — What is the downstream task for which the dataset was created?
- `not_intended_for` — Explicit exclusions or use cases the data should not be used for

### Ethics
- `privacy_method` — Privacy-preserving techniques used (e.g., differential privacy, k-anonymity, synthetic-from-scratch, etc.)

### Reproducibility
- `code_repository` — Link to the code repository or generation scripts
- `environment` — Software environment, dependencies, or Docker image used
